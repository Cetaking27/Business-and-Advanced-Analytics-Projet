3. Compare churned and non-churned customers
</span>
<span  style="font-family:'New time Roman';">

Based on Churn Label group :
- **Account Length (in months)**
- **Avg Monthly GB Download**

```python
numerical_variables = ["Account Length (in months)", "Avg Monthly GB Download", "Age", "Total Charges"]

for col in numerical_variables:
    if col in churnDataset.columns:
        print(f"--- {col} ---")
        stats = churnDataset.groupby('Churn Label')[col].agg(['mean', 'median', 'std'])
        print(stats)
        print("\n")
```

<span  style="font-family:'New time Roman';">

 Grouping our customer by  those who churned (**Churn Label** = *YES*) and who stayed (**Churn Label** = *NO*), we find out that :
- Customers who churned had shorter account durations, averaging `18.07 months`, compared with `37.58 months` for those who stayed. Median durations were `10` and `37 months`, respectively.
- Customers who churned downloaded slightly more data, averaging `7.19 GB per month`, compared with `6.52 GB` for those who stayed.
- Customers who churned were older on average, at `50.46 years`, compared with `46.34 years` for those who stayed.
- Customers who churned had lower average total charges `(761.42)` than those who stayed `(1,202.12)`. This is consistent with their shorter account durations.

```python
catergorical_var = ["Churn Label","State" ,"Contract Type", "Payment Method"]
for col in catergorical_var:
    if col in churnDataset.columns:
        count =  churnDataset[col].value_counts()
        percentage = churnDataset[col].value_counts(normalize=True)
        table = pd.DataFrame({
            "Count": count,
            "Percentage": round(percentage*100, 2)
        })
        print(f"--------- {col} ------------")
        print(table.to_string())
        print(f"------------------------")
        print(f"The : {percentage.idxmax()}, is the maximum count with percentage = {round(percentage.max()*100,2)} %")
        print(f"------------------------")
        print("\n")
```
